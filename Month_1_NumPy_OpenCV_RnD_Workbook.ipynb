### 1D
- arr[i]
- arr[start:end]
- arr[:n]
- arr[-n:]
- arr[condition]

### 2D
- arr[row, col]
- arr[row]
- arr[:, col]
- arr[row, :]
- arr[r1:r2, c1:c2]

### 3D
- arr[depth, row, col]
- arr[:, row, col]
- arr[depth, :, :]
- arr[:, :, col]

### Boolean
- arr[arr > 80]
- arr[(arr > 70) & (arr < 90)]
- arr[(arr > 90) | (arr < 60)]

```python
# ============================================================
# NUMPY INDEXING DRILL
# ============================================================

import numpy as np

# ------------------------------------------------------------
# 1D ARRAY
# ------------------------------------------------------------

arr1 = np.arange(1, 21)

print("arr1:")
print(arr1)

# TODO 1
# İlk 5 elemanı seç.
print(arr1[:5])

# TODO 2
# Son 4 elemanı seç.
print(arr1[-4:])

# TODO 3
# Index 5 ile index 12 arasındaki elemanları seç.
# 5 dahil, 12 hariç.
print(arr1[5:12])

# TODO 4
# Baştan başlayarak 2'şer atlayarak elemanları seç.
print(arr1[::2])

# TODO 5
# 10'dan büyük elemanları seç.
print(arr1[arr1>10])

# TODO 6
# 5 ile 15 arasındaki elemanları seç.
# 5 ve 15 dahil.
print(arr1[(arr1>=5)&(arr1<=15)])


```

```python
# ------------------------------------------------------------
# 2D ARRAY
# ------------------------------------------------------------

arr2 = np.arange(1, 25).reshape(4, 6)

print("\narr2:")
print(arr2)

# TODO 7
# İlk satırı seç.
arr2[:5]

# TODO 8
# Son satırı seç.
arr2[-1:]

# TODO 9
# 2. satırı seç. (index = 1)
arr2[1,:]

# TODO 10
# İlk sütunu seç.
arr2[:,1]

# TODO 11
# Son sütunu seç.
arr2[:,-1]

# TODO 12
# İlk 2 satırı seç.
arr2[:2,:]

# TODO 13
# 2. ve 3. sütunları seç. (index 1 ve 2)
arr2[:,1:3]

# TODO 14
# İlk 3 satır + ilk 4 sütunu seç.
arr2[:3,:4]

# TODO 15
# 2. satırdan son satıra kadar + son 3 sütunu seç.
arr2[2:,-3:]

# TODO 16
# Çift indexli satırları seç.
##even=lambda x: x if x%2==0 else None
arr2[0::2,:]

# TODO 17
# Tek indexli sütunları seç.
arr2[1::2,:]

# TODO 18
# 15'ten büyük tüm değerleri seç.
arr2[arr2>15]
```

```python
# ------------------------------------------------------------
# 2D BOOLEAN MASKING
# ------------------------------------------------------------

# TODO 19
# 10'dan büyük VE 20'den küçük değerleri seç.
arr2[(arr2>10)&(arr2<20)]

# TODO 20
# 5'ten küçük VEYA 20'den büyük değerleri seç.
arr2[(arr2<5) | (arr2>20)]
```

```python
# ------------------------------------------------------------
# SHAPE DRILL
# ------------------------------------------------------------

# Aşağıdakilerin shape'lerini ÖNCE TAHMİN ET,
# sonra print ederek kontrol et.

# TODO 21
# arr2[0]
# shape (1,1)
arr2[0].shape
#(6,)

# TODO 22
# arr2[0:1]
#shape (2,2)
arr2[0:1].shape
#(1, 6)

# TODO 23
# arr2[:, 0]
#shape (4,1)
arr2[:, 0].shape
#(4,)

# TODO 24
# arr2[:, 0:1]
#shape (4,0)
arr2[:, 0:1].shape
#(4, 1)

# TODO 25
# arr2[1:3, 2:5]
#shape (2,3)
arr2[1:3, 2:5].shape
#(2,3)



"""
[0]    → tek eleman / tek satır / tek sütun → boyut düşebilir
[0:1]  → slice → boyut korunur
"""
```

```python
# ------------------------------------------------------------
# 3D ARRAY
# ------------------------------------------------------------

arr3 = np.arange(1, 61).reshape(3, 4, 5)

print("\narr3 shape:")
print(arr3.shape)

# Shape:
# depth = 3
# rows  = 4
# columns = 5

# TODO 26
# İlk 2D bloğu seç.
arr3[0,:,:]

# TODO 27
# İkinci depth bloğunu seç.
arr3[1,:,:]


# TODO 28
# Bütün depth'lerde ilk satırı seç.
arr3[:,0,:]

# TODO 29
# Bütün depth'lerde ilk sütunu seç.
arr3[:,:,0]

# TODO 30
# Tüm depth'lerde 2. ve 3. sütunları seç.
arr3[:,:,1:3]

# TODO 31
# İlk 2 depth'i seç.
arr3[0:2,:,:]

# TODO 32
# İlk 2 depth + ilk 2 satır + son 3 sütunu seç.
arr3[0:2,0:2,-3:]
```

```python
# ------------------------------------------------------------
# FINAL SHAPE CHALLENGE
# ------------------------------------------------------------

# Her biri için:
# 1) Önce shape'i tahmin et
# 2) Sonra print et

# TODO 33
result1 = arr3[:, :, 0]
#shape (3,4,)
##(3, 4)

# TODO 34
result2 = arr3[:, 1, :]
#shape (3,1,5)
##(3,5)

# TODO 35
result3 = arr3[0:2, :, 1:4]
#shape (2,4,3)
##(2, 4, 3)

# TODO 36
result4 = arr3[:, 0:2, :]
#shape (3,2,5)
##(3,2,5)

# TODO 37
result5 = arr3[1:, 1:3, 2:]
#shape (2,2,3)
##(2,2,3)

print(result1.shape)
print(result2.shape)
print(result3.shape)
print(result4.shape)
print(result5.shape)
```
